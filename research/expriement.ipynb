```python
file_path = "data/SDG.pdf"
loader = PyPDFLoader(file_path)
data = loader.load()
```

```python
question_gen = ""
for page in data:
    question_gen += page.page_content
```

```python
question_gen
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter_ques_gen = RecursiveCharacterTextSplitter(
    chunk_size=10000,
    chunk_overlap=200
)
```

```python
chunk_ques_gen = splitter_ques_gen.split_text(question_gen)
```

```python
chunk_ques_gen
```

```python
len(chunk_ques_gen)
```

```python
from langchain.docstore.document import Document
```

```python
document_ques_gen = [Document(page_content = t) for t in chunk_ques_gen]
```

```python
document_ques_gen
```

```python
type(document_ques_gen[0])
```

```python
splitter_ans_gen = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)
```

```python
document_answer_gen = splitter_ans_gen.split_documents(
    document_ques_gen
)
```

```python
document_answer_gen
```

```python
len(document_answer_gen)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm_ques_gen_pipeline = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash", 
    google_api_key=GOOGLE_API_KEY,
    temperature=0.3
)
```

```python
prompt_template = """
You are an expert at creating questions based on coding materials and documentation.
Your goal is to prepare a coder or programmer for their exam and coding tests.
You do this by asking 10 questions about the text below:

------------
{text}
------------

Create questions that will prepare the coders or programmers for their tests.
Make sure not to lose any important information.

QUESTIONS:
"""
```

```python
from langchain.prompts import PromptTemplate
```

```python
PROMPT_QUESTIONS = PromptTemplate(template=prompt_template, input_variables=['text'])
```

```python
refine_template = ("""
You are an expert at creating practice questions based on coding material and documentation.
Your goal is to help a coder or programmer prepare for a coding test.
We have received some practice questions to a certain extent: {existing_answer}.
We have the option to refine the existing questions or add new ones.
(only if necessary) with some more context below.
------------
{text}
------------

Given the new context, refine the original questions in English.
If the context is not helpful, please provide the original questions.
QUESTIONS:
"""
)
```

```python
REFINE_PROMPT_QUESTIONS = PromptTemplate(
    input_variables=["existing_answer", "text"],
    template=refine_template,
)
```

```python
from langchain.chains.summarize import load_summarize_chain
```

```python
ques_gen_chain = load_summarize_chain(llm = llm_ques_gen_pipeline, 
                                          chain_type = "refine", 
                                          verbose = True, 
                                          question_prompt=PROMPT_QUESTIONS, 
                                          refine_prompt=REFINE_PROMPT_QUESTIONS)
```

```python
ques = ques_gen_chain.run(document_ques_gen)

print(ques)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/embedding-001"  # Gemini's embedding model
)
```

```python
from langchain.vectorstores import FAISS
```

```python
vector_store = FAISS.from_documents(document_answer_gen, embeddings)
```

```python
llm_answer_gen = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    google_api_key=GOOGLE_API_KEY,
    temperature=0.1
)
```

```python
ques
```

```python
ques_list = ques.split("\n")
```

```python
ques_list
```

```python
from langchain.chains import RetrievalQA
```

```python
answer_generation_chain = RetrievalQA.from_chain_type(llm=llm_answer_gen, 
                                               chain_type="stuff", 
                                               retriever=vector_store.as_retriever())
```

```python
# Answer each question and save to a file
with open("answers.txt", "w") as f:
    for question in ques_list:
        if question.strip():  # Skip empty questions
            print("Question:", question)
            try:
                answer = answer_generation_chain.run(question)
                print("Answer:", answer)
            except Exception as e:
                print("Error answering question:", e)
                answer = "Unable to generate answer due to an error."
            print("--------------------------------------------------\n\n")
            
            # Write to file
            f.write(f"Question: {question}\n")
            f.write(f"Answer: {answer}\n")
            f.write("--------------------------------------------------\n\n")

```
